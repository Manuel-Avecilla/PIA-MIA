3. Aplicar las operaciones aritméticas básicas (suma, resta, multiplicación y división) sobre objetos Series.

```python
s003 = pd.Series([1,2,3])

s003_suma = s003 + 1
print(s003_suma)

s003_resta = s003 - 0.5
print(s003_resta)

s003_mult = s003 * 3
print(s003_mult)

s003_div = s003 / 4
print(s003_div)
```
