33. Elimina los outliers dentro de la lista del ejercicio de ejemplo de Pandas.

```python
# Serie del ejercicio de ejemplo de Pandas:

# Temperaturas promedio diarias en grados Celsius para un mes (30 días)
temps = [22, 21, 23, 24, 26, 27, 29, 30, 33, 35, 34, 31, 30, 28, 27, 25, 24, 45, 42, 39,40 , 50,23, 22, 21, 20, 22, 23, 25, 28, 29, 31, 32, 33, 34,np.nan,45]

# Crear una Serie de Pandas
temperature_series = pd.Series(temps, name="Temperatura")
# print("Serie de temperaturas: \n", temperature_series)

# Eliminar los outliers
Q1 = temperature_series.quantile(0.25)
Q3 = temperature_series.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print("Límites del rango intercuartílico: \n", lower_bound, upper_bound)

temperature_series = temperature_series[(temperature_series >= lower_bound) & (temperature_series <= upper_bound)]
print("Serie sin outliers: \n", temperature_series)
```

34. En una Serie de Pandas donde se registran las ventas mensuales de un producto durante años, utiliza el método group by para calcular las ventas medias en cada mes.

```python
fechas = pd.date_range(start="2021-01-01", periods=36, freq="MS") # MS = inicio de mes
ventas = pd.Series(np.random.randint(1000, 5000, size=36), index=fechas, name="Ventas")

ventas_medias_mes = ventas.groupby(ventas.index.month).mean()

print(ventas_medias_mes)
```
