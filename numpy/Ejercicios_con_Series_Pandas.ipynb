```python
s003 = pd.Series([1,2,3])

s003_suma = s003 + 1
print(s003_suma)

s003_resta = s003 - 0.5
print(s003_resta)

s003_mult = s003 * 3
print(s003_mult)

s003_div = s003 / 4
print(s003_div)
```

13. Obtener la desviación estándar y el promedio de un conjunto de datos de una serie.

```python
s013 = pd.Series(np.random.randint(1, 7, size=20))
print(s013)

promedio = s013.describe().mean()
print("promedio: ",promedio)

desviacion = s013.describe().std()
print("Desviacion: ",desviacion)
```

16. Obtener los elementos pares e impares de un objeto Series numérico.

```python
s016 = pd.Series([1,2,3,4,5])

s016_pares = s016[s016 % 2 == 0]
print(s016_pares)

s016_impares = s016[s016 % 2 != 0]
print(s016_impares)
```

18. Comprobar si una serie contiene valores NaN.

```python
d018 = {"a":1,"b":5,"d":7,"f":8}

s018 = pd.Series(d018,["a","b","c","d","e","f"])

print(s018)

print("¿ La serie contiene NaN ? -> ",s018.hasnans)
```
