## Ejercicio 11

Crea una función llamada f que acepte como parámetro el número y que retorne lo
siguiente:

- Si es valor está en el rango (-inf,3] que retorne 0

- Si es valor está en el rango [3,5] que retorne 4

- Si es valor está en el rango [5,6] que retorne 5

- Si es valor está en el rango [6,7] que retorne 6

- Si es valor está en el rango [7,9] que retorne 8

- Si es valor está en el rango [9,+inf) que retorne 10


```python
def f(num):
    match num:
        case x if x <= 3:
            return 0
        case x if x <= 5:
            return 4
        case x if x <= 6:
            return 5
        case x if x <= 7:
            return 6
        case x if x <= 9:
            return 8
        case _:
            return 10
```

- Crea ahora el vector de numpy [1,5,4,7,3,9,8,6] y aplícale la función f

```python
vector = np.array([1, 5, 4, 7, 3, 9, 8, 6])
f_vec = np.vectorize(f)
print(f_vec(vector))
```

## Ejercicio 18

Carga con numpy el fichero wine.csv que contiene una matriz de números
separados por coma.
La matriz contiene una serie de columnas con características del vino y la última
columna indica el tipo del vino.

- Crea una función en python llamada imprimir_datos y tendrá como parámetro
  una matriz de numpy de forma que:
  - Imprima cuantas características tenemos de cada tipo de vino.
  - Imprima cuantas muestra de vino tenemos
  - Imprima cuantos tipos distintos de vino tenemos y cuales son
  - Imprima los valores máximos y mínimos de cada característica
  - Retorna una nueva matriz llamada x solo con las características y un
    nuevo array llamado y solo con los tipos de vino
- Llama a la función imprimir_datos con los datos de wine.csv

```python
def imprimir_datos(matriz):
    x = matriz[:, :-1]
    y = matriz[:, -1]
    
    num_caracteristicas = x.shape[1]
    print(f"Número de características por vino: {num_caracteristicas}")
    
    num_muestras = matriz.shape[0]
    print(f"Número total de muestras de vino: {num_muestras}")
    
    tipos = np.unique(y)
    print(f"Número de tipos distintos de vino: {len(tipos)}")
    print(f"Tipos de vino: {tipos}")
    
    for tipo in tipos:
        conteo = np.sum(y == tipo)
        print(f"  - Tipo {int(tipo)}: {conteo} muestras ({num_caracteristicas} características)")
        
    maximos = np.max(x, axis=0)
    minimos = np.min(x, axis=0)
    print("\nValores mínimos y máximos de cada característica:")
    for i in range(num_caracteristicas):
        print(f"  Característica {i+1}: Mínimo = {minimos[i]:.4f}, Máximo = {maximos[i]:.4f}")
        
    return x, y

datos_wine = np.loadtxt("wine.csv", delimiter=",")
x, y = imprimir_datos(datos_wine)
```

## Ejercicio 19

Tienes los datos de las ventas diarias de una tienda durante un año. Los datos
incluyen ventas anormalmente altas debido a días festivos o promociones. El
objetivo es limpiar los datos eliminando estos valores atípicos y analizar la tendencia
general de ventas.

```python
np.random.seed(0)

# Ventas diarias simuladas con media 500 y desviación estándar 50

sales_data = np.random.normal(500, 50, 365)

# Ventas atípicas por promociones
sales_data[[50, 120, 180, 250, 300]] = [1500, 1400, 1600, 1700, 1800] 
```

- Detecta y elimina los valores atípicos (aquellos que están fuera de 3
desviaciones estándar del promedio).

```python
media = np.mean(sales_data)
desv = np.std(sales_data)

atipicos = sales_data[np.abs(sales_data - media) > 3 * desv]
print("Valores atípicos detectados:", atipicos)

sales_clean = sales_data[np.abs(sales_data - media) <= 3 * desv]
print(f"Número de ventas antes de limpiar: {len(sales_data)}")
print(f"Número de ventas después de limpiar: {len(sales_clean)}")
```

- Calcula el promedio, la mediana y la desviación estándar de las ventas
después de limpiar los datos.

```python
promedio = np.mean(sales_clean)
mediana = np.median(sales_clean)
desviacion = np.std(sales_clean)

print(f"Promedio de ventas: {promedio:.2f}")
print(f"Mediana de ventas: {mediana:.2f}")
print(f"Desviación estándar de ventas: {desviacion:.2f}")
```

- Genera una vista de las ventas de cada trimestre (3 primeros meses, 3
siguientes, etc.).

```python
# Vistas de las ventas de cada trimestre
q1 = sales_data[0:90]
q2 = sales_data[90:181]
q3 = sales_data[181:273]
q4 = sales_data[273:365]

print("Ventas Trimestre 1 (Q1, 90 días):", q1[:5], "...")
print("Ventas Trimestre 2 (Q2, 91 días):", q2[:5], "...")
print("Ventas Trimestre 3 (Q3, 92 días):", q3[:5], "...")
print("Ventas Trimestre 4 (Q4, 92 días):", q4[:5], "...")

# O dividiendo en 4 trimestres con np.array_split:
trimestres = np.array_split(sales_data, 4)
for i, t in enumerate(trimestres, 1):
    print(f"Trimestre {i}: {len(t)} días, promedio = {np.mean(t):.2f}")
```

- Encuentra los 5 días con las ventas más bajas después de la limpieza de los
datos.

```python
indices_mas_bajas = np.argsort(sales_clean)[:5]
ventas_mas_bajas = sales_clean[indices_mas_bajas]

print("Los 5 días con las ventas más bajas tras la limpieza:")
for puesto, (idx, venta) in enumerate(zip(indices_mas_bajas, ventas_mas_bajas), 1):
    print(f"{puesto}º - Posición/Día {idx}: {venta:.2f}")
```

## Ejercicio 20

Tienes un conjunto de datos que contiene información demográfica de varias
personas, incluyendo edad, ingresos anuales, y nivel de educación. El objetivo es
realizar un análisis y encontrar patrones significativos entre estas variables.

- Para calcular la correlación utiliza la función corrcoef.
- Para encontrar el percentil 90 utiliza la función percentile.

```python
np.random.seed(3)

# Edades de 200 personas
ages = np.random.randint(18, 70, size=200)

# Ingresos anuales en un rango normal alrededor de 50,000
income = np.random.normal(50000, 15000, size=200) 

# Años de educación formal
education_years = np.random.randint(0, 20, size=200) 
```

- Calcula la correlación entre los años de educación y los ingresos.

```python
matriz_correlacion = np.corrcoef(education_years, income)
print("Matriz de correlación:")
print(matriz_correlacion)

correlacion = matriz_correlacion[0, 1]
print(f"\nCorrelación entre años de educación e ingresos: {correlacion:.4f}")
```

- Divide a las personas en grupos de edades (18-30, 31-45, 46-60, 61+) y
calcula el ingreso promedio por grupo.

```python
g_18_30 = income[(ages >= 18) & (ages <= 30)]
g_31_45 = income[(ages >= 31) & (ages <= 45)]
g_46_60 = income[(ages >= 46) & (ages <= 60)]
g_61_plus = income[ages >= 61]

print(f"Grupo 18-30 años ({len(g_18_30)} personas): Ingreso promedio = {np.mean(g_18_30):.2f}")
print(f"Grupo 31-45 años ({len(g_31_45)} personas): Ingreso promedio = {np.mean(g_31_45):.2f}")
print(f"Grupo 46-60 años ({len(g_46_60)} personas): Ingreso promedio = {np.mean(g_46_60):.2f}")
print(f"Grupo 61+ años   ({len(g_61_plus)} personas): Ingreso promedio = {np.mean(g_61_plus):.2f}")
```

- Encuentra el percentil 90 de ingresos y determina cuántas personas ganan
por encima de ese valor.

```python
p90 = np.percentile(income, 90)
personas_encima = np.sum(income > p90)

print(f"Percentil 90 de ingresos: {p90:.2f}")
print(f"Número de personas que ganan por encima del percentil 90: {personas_encima}")
```

- Agrupa los datos en función del número de años de educación y calcula el
promedio de ingresos para cada grupo.
