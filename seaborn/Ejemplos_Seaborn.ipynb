```python
import seaborn as sns
import matplotlib.pyplot as plt

# Cargar el dataset tips
tips = sns.load_dataset("tips")

# Configurar la figura y los subplots
plt.figure(figsize=(12, 10))

# Gráfica 1: Distribución de propinas (tip)
plt.subplot(2, 2, 1)
sns.histplot(data=tips, x="tip", kde=True, color="skyblue")
plt.title("Distribución de Propinas")
plt.xlabel("Propina")
plt.ylabel("Frecuencia")

# Gráfica 2: Diagrama de caja de las propinas por día
plt.subplot(2, 2, 2)
sns.boxplot(data=tips, x="day", y="tip", palette="muted")
plt.title("Propinas por Día")
plt.xlabel("Día")
plt.ylabel("Propina")

# Gráfica 3: Total de la cuenta promedio por tamaño de grupo
plt.subplot(2, 2, 3)
sns.barplot(data=tips, x="size", y="total_bill", palette="viridis")
plt.title("Total de la Cuenta por Tamaño de Grupo")
plt.xlabel("Tamaño de Grupo")
plt.ylabel("Total de la Cuenta")

# Gráfica 4: Relación entre total de cuenta y propina
plt.subplot(2, 2, 4)
sns.scatterplot(data=tips, x="total_bill", y="tip", hue="time", style="sex", palette="deep")
plt.title("Relación entre Total y Propina")
plt.xlabel("Total de la Cuenta")
plt.ylabel("Propina")

# Ajustar diseño
plt.tight_layout()
plt.show()
```

```python
tips = sns.load_dataset("tips")


sns.catplot(x = "day", data = tips, kind = "count");
```

```python
groupedvalues = tips[['tip','day']].groupby('day').sum().reset_index()
g = sns.barplot(x = 'day', y = 'tip', data = groupedvalues)
for index, row in groupedvalues.iterrows():
    g.text(row.name, row.tip+5, round(row.tip), color = 'black', ha = "center")
```
