### 37. Seleccionar registros donde los valores de una columna estén en un rango.

```python
df_037= pd.DataFrame({
    'Producto': ['A', 'B', 'C', 'D'],
    'Precio': [10.5, 20.0, 15.75, 30.0],
    'Cantidad': [100, 200, 150, 300]
})

df_037.where((df_037['Precio'] > 15.0) & (df_037['Cantidad'] > 150)).dropna()
```
