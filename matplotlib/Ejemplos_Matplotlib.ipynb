# Gráfico de líneas básico

```python
import matplotlib.pyplot as plt
import numpy as np


# 1. Crear gráficos básicos
# Vamos a crear un gráfico de línea simple.

# Datos de ejemplo
x = np.linspace(0, 10, 100)  # Crear un arreglo de 100 valores entre 0 y 10
y = np.sin(x)  # La función seno de x

# Crear el gráfico
plt.plot(x, y)  # Dibuja una línea conectando los puntos de x, y
plt.title("Gráfico de Seno")  # Título del gráfico
plt.xlabel("Eje X")  # Etiqueta para el eje X
plt.ylabel("Eje Y")  # Etiqueta para el eje Y
plt.grid(True)  # Añadir una cuadrícula al fondo
plt.show()  # Mostrar el gráfico

```

```python
from matplotlib.ticker import MultipleLocator, FormatStrFormatter
fig=plt.figure()

ax1=plt.axes()
ax1.yaxis.set_major_locator(MultipleLocator(5))
ax1.yaxis.set_minor_locator(MultipleLocator(1))
#ax1.yaxis.set_minor_formatter(FormatStrFormatter('%0.4f'))

ax1.plot([1,2,3,5,10],"o--r")
ax2=plt.axes([0.25,0.25,0.5,0.5])
ax2=plt.plot([1,2,5])
plt.show()
```

```python
#plt.figure()
plt.plot([1,2,3,5,10],"o--r",label="Ene")
plt.plot([-1,5,5,56],color="SteelBlue",label="Feb")
plt.plot([-2,7,12,56],color="green",label="Feb")
plt.plot([2],[12],color='g',marker="*",markersize=12)
plt.legend(loc=9)
plt.show()
```

# Algunos aspectos de personalización

```python
# 2. Personalizar el gráfico
# Vamos a modificar el gráfico anterior agregando color, estilo de línea y marcadores.

plt.plot(x, y, color='red', linestyle='--', marker='o')  # Establecer color, estilo y marcadores
plt.title("Gráfico de Seno Personalizado")  # Título del gráfico
plt.xlabel("Eje X")  # Etiqueta para el eje X
plt.ylabel("Eje Y")  # Etiqueta para el eje Y
plt.grid(True)  # Cuadrícula
plt.show()
```

# Subplots

```python
# 3. Subgráficos (subplots)
# Vamos a crear dos gráficos en una misma ventana.

fig, axs = plt.subplots(1, 2)  # Crear 1 fila, 2 columnas de subgráficos
axs[0].plot(x, y, 'b')  # Primer subgráfico con línea azul
axs[0].set_title("Gráfico 1: Seno")  # Título del primer subgráfico
axs[0].set_xlabel("X")  # Etiqueta para el eje X
axs[0].set_ylabel("Y")  # Etiqueta para el eje Y
axs[1].plot(x, np.cos(x), 'g')  # Segundo subgráfico con línea verde
axs[1].set_title("Gráfico 2: Coseno")  # Título del segundo subgráfico
axs[1].set_xlabel("X")  # Etiqueta para el eje X
axs[1].set_ylabel("Y")  # Etiqueta para el eje Y

plt.tight_layout()  # Ajusta los subgráficos para que no se sobrepongan
plt.show()
```

# Diagrama de Caja y Bigotes

```python
# 6. Diagrama de caja y bigotes
# Generando datos de ejemplo
np.random.seed(10)  # Fijar la semilla para resultados reproducibles
data = np.random.normal(loc=0, scale=1, size=100)  # Datos normalmente distribuidos
data[25]=3
# Crear el diagrama de caja
plt.boxplot(data)

# Añadir título y etiquetas
plt.title("Diagrama de Caja y Bigotes")
plt.xlabel("Valor")
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

```python
import pandas as pd

# 6. Diagrama de caja y bigotes por categoría
# Crear datos de ejemplo
np.random.seed(10)  # Para reproducibilidad
categorias = ['A', 'B', 'C']  # Categorías
data = {
    'Categoria': np.random.choice(categorias, 200),  # 200 puntos aleatorios entre 'A', 'B', 'C'
    'Valor': np.random.randn(200)  # 200 valores aleatorios generados a partir de una distribución normal
}

# Crear un DataFrame
df = pd.DataFrame(data)

# Organizar los datos por categoría para la creación del boxplot
categorias_unicas = df['Categoria'].unique()  # Obtener categorías únicas
valores_por_categoria = [df[df['Categoria'] == categoria]['Valor'] for categoria in categorias_unicas]

# Crear el diagrama de caja
plt.figure(figsize=(8, 6))
plt.boxplot(valores_por_categoria, label=categorias_unicas)

# Títulos y etiquetas
plt.title("Diagrama de Caja por Categoría")
plt.xlabel("Categoría")
plt.ylabel("Valor")
plt.grid(True)

# Mostrar el gráfico
plt.show()
```

# Representar datos en un Dataframe

```python
import pandas as pd
import matplotlib.pyplot as plt

# Crear un DataFrame de ejemplo
data = {
    'Mes': ['Enero', 'Febrero', 'Marzo', 'Abril', 'Mayo'],
    'Producto A': [250, 270, 290, 310, 330],
    'Producto B': [200, 220, 240, 260, 280],
    'Producto C': [150, 180, 210, 240, 270]
}

df = pd.DataFrame(data)

# Configurar el índice como el mes
df.set_index('Mes', inplace=True)

# Representar los datos como un gráfico de líneas
plt.figure(figsize=(10, 6))
plt.plot(df.index, df['Producto A'], marker='o', label='Producto A')
plt.plot(df.index, df['Producto B'], marker='s', label='Producto B')
plt.plot(df.index, df['Producto C'], marker='^', label='Producto C')

# Añadir etiquetas y título
plt.title('Ventas mensuales por producto', fontsize=14)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Ventas', fontsize=12)

# Añadir una leyenda
plt.legend()

# Mostrar la cuadrícula
plt.grid()

# Mostrar el gráfico
plt.show()
```
